 🧮 Operaciones matemáticas y lógicas
 Qué deben saber:
 * Operadores básicos: +, -, *, /, **, %, //
 * Comparaciones: >, <, >=, <=, ==, !=
 * Condicionales con if y else


```python
# Ejemplo operadores básicos +
n1=10
n2=20
suma=n1+n2
print(suma)
```

```python
# Ejemplo operadores básicos -
n1=10
n2=20
resta=n1-n2
print(resta)
```

```python
# Ejemplo operadores básicos *
n1=10
n2=20
mult=n1*n2
print(mult)
```

```python
# Ejemplo operadores básicos /
n1=10
n2=20
div=n1/n2
print(div)
```

```python
# Ejemplo operadores básicos **
n1=2
n2=3
exp=n1**n2
print(exp)
```

```python
# Ejemplo operadores básicos %
n1=25
n2=10
modulo=n1%n2
print(modulo)
```

```python
# Ejemplo operadores básicos //
n1=25
n2=10
modulo=n1//n2
print(modulo)
```

```python
# Comparaciones: >
n1=10
n2=20
comp=n1>n2
print(comp)
```

```python
# Comparaciones: <
n1=10
n2=20
comp=n1<n2
print(comp)
```

```python
# Comparaciones: >=
n1=10
n2=20
comp=n1>=n2
print(comp)
```

```python
# Comparaciones: <=
n1=10
n2=20
comp=n1<=n2
print(comp)
```

```python
# Comparaciones: !=
n1=10
n2=20
comp=n1!=n2
print(comp)
```

```python
# Comparaciones: ==
n1=10
n2=20
comp=n1==n2
print(comp)
```

```python
# Ejemplo
edad= 17
if edad >= 18:
  print("Eres mayor de edad")
else:
  print("Eres menor de edad")
```

### Estructuras de repetición
Qué deben saber:
* uso for y while

```python
# Ejemplo for
for i in range(5):
  print("Número ",i)
```

```python
# Ejemplo while
contador=0
while contador<5:
  print("Número ",contador)
  contador=contador+1
```
